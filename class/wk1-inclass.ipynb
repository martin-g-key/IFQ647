```python
import matplotlib.pyplot as plt
```

```python
text = "Tropical fish are generally those fish found in aquatic tropical environments around the world, including both freshwater and saltwater species. Fishkeepers often keep tropical fish in freshwater and saltwater aquariums. Fishkeepers often keep tropical fish in freshwater and saltwater aquariums. "
```

```python
def splitter(text_input=text):

    doc = {}
    word_count = 0 

    ## assumes text is in one line -- ie. not extacted from json or xml 
    for term in text.split():
        word_count += 1
        term = term.lower()
        if term.isalpha():
            try:
                doc[term] += 1
            except KeyError:
                doc[term] = 1

    # reverse order the doc
    doc_list = doc.items()
    doc_list = sorted(doc_list, key=lambda x: x[1], reverse=True)

    # plot the data 
    x, y = zip(*doc_list)
    plt.plot(x,y)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    return doc, word_count, doc_list
```
