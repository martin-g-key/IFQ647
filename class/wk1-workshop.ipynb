```python
# Task 1

import os

def task_1(folder, filename):
    ## A: load xml file
    filepath = os.path.join(folder,filename)
    with open(filepath, "r", encoding="utf-8") as f:
        print(f"{filename} opened.")

        f_ = f.readlines()
        start_end=False
        news = "" # blank string to add text to 
        word_count = 0 

        for line in f_:
            line = line.strip()
            if (start_end == False): 
                if line.startswith("<newsitem "): 
                    for part in line.split():
                        if part.startswith("itemid="):
                            docid = part.split("=")[1].split("\"")[1]
                            break 
                if line.startswith("<text>"):
                    start_end=True 
            elif line.startswith("</text>"):
                break
            else:
                # remove paragraph tags
                line = line.replace("<p>", "").replace("</p>", "")
                news = news + line + "\n"
                for term in line.split():
                    word_count += 1 


    f.close()
    print(f"{filename} closed.")

    print(f"Item id: {docid}")
    print(f"Word count: {word_count}")

    return news, docid


```

```python
folder = "/Users/martin/Documents/ifq647_working/files/"
file = "741299newsML.xml"

news, docid = task_1(folder,file)
```

```python
# task 2 
# takes news from task 1 as input. 

import string

def text_parser(folder, input, stops):

    ## Define stop words and store in list 
    filepath = os.path.join(folder, stops)
    with open(filepath, 'r', encoding="utf-8") as f:
        print(f"{stops} is open")
        stop_words = f.read().split(',')
    f.close()
    print(f"{stops} is closed")

    ## select sentence
    ## not required - just use news string from task 1 
    text = input
    #print(text)

    ## exclude all tags
    text = text.replace("<p>", "").replace("</p>", "")
    #print(text)

    ## discard punctuation
    text = text.translate(str.maketrans(string.punctuation, ' '*len(string.punctuation)))
    #print(text)

    ## discard numbers 
    text = text.translate(str.maketrans('', '', string.digits))
    #print(text)

    ## turn into dictionary and remove stop words 
    text_dict = {}
    for term in text.split(): 
        if term not in stop_words and len(term) >= 2:
            try: 
                text_dict[term] += 1 
            except KeyError: 
                text_dict[term] = 1


    # reverse order the term list
    term_list = text_dict.items()
    term_list = sorted(term_list, key=lambda x: x[1], reverse=True)

    #print(term_list)

    print(len(text),text_dict )
```
